```python
#Encapsulation in Python

class Employee:
    def __init__(self, name, salary):
        self.name = name          # public
        self.__salary = salary    # private

    def show_salary(self):
        print("Salary:", self.__salary)

emp = Employee("Robert", 60000)
print(emp.name)          # Public accessible
emp.show_salary()        # Accessing private correctly
print(emp.__salary)    # Error: Not accessible directly
```

```python
class person:
    def __init__(self):
        self.__age = 0 #privete variable
    def get__age(self):
        return self.__age
    def set__age(self, age):
        self.__age = age
p = person()
p.set__age(21) #set value
print(p.get__age()) #get value
```

```python
#Polymorphism in Python

class Calculator:
    def multiply(self, a=1, b=1, *args):
        result = a * b
        for num in args:
            result *= num
        return result

# Create object
calc = Calculator()

# Using default arguments
print(calc.multiply())            
print(calc.multiply(4))           

# Using multiple arguments
print(calc.multiply(2, 3))       
print(calc.multiply(2, 3, 4))
```

```python
from abc import ABC, abstractmethod
#prent class
class person(ABC):
    def __init__(self):
        print("calling constractor")
    @abstractmethod
    def fun(self):
        print("fun is called")
#child class
class man(person):
    def fun(self):
        super().__init__
        super().fun()
        #print("man fun is called")
        return print("man fun is called")

m = man()
m.fun()

    
```
