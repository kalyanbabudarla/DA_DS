# PART 1: 📊 Pandas Series
A **Series** is a one-dimensional labeled array. Think of it as a column in a spreadsheet.

## 1.1 Creating a Series

```python
# From a list
s1 = pd.Series([10, 20, 30, 40, 50])
print('From list:')
print(s1)
print()
```

```python
# From a list with custom index
s2 = pd.Series([85, 90, 78, 92, 88],
               index=['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
               name='Scores')
print('With custom index:')
print(s2)
```

```python
# From a dictionary
s3 = pd.Series({'Mon': 120, 'Tue': 95, 'Wed': 110, 'Thu': 130, 'Fri': 105})
print('From dictionary:')
print(s3)
```

```python
# From a scalar (broadcasts the value)
s4 = pd.Series(7, index=['a', 'b', 'c', 'd'])
print('From scalar:')
print(s4)
```

## 1.2 Series Attributes

```python
print('Values:', s2.values)
print('Index:', s2.index.tolist())
print('Name:', s2.name)
print('dtype:', s2.dtype)
print('Shape:', s2.shape)
print('Size:', s2.size)
print('ndim:', s2.ndim)
```

## 1.3 Accessing Elements

```python
# By label
print('Alice score:', s2['Alice'])

# By position
print('First score:', s2.iloc[0])

# Slicing
print('\nFirst 3 students:')
print(s2.iloc[:3])

# Multiple labels
print('\nAlice & Diana:')
print(s2[['Alice', 'Diana']])
```

## 1.4 Series Operations

```python
# Arithmetic
print('Scores + 5 bonus:')
print(s2 + 5)
print()

# Boolean filtering
print('Students scoring above 88:')
print(s2[s2 > 88])
```

```python
# Useful methods
print('Sum:', s2.sum())
print('Mean:', s2.mean())
print('Max:', s2.max())
print('Min:', s2.min())
print('Std Dev:', s2.std().round(2))
print('Median:', s2.median())
```

```python
# Sorting
print('Sorted by value:')
print(s2.sort_values(ascending=False))

print('\nSorted by index:')
print(s2.sort_index())
```

## 1.5 Handling Missing Values in Series

```python
s_missing = pd.Series([10, None, 30, np.nan, 50], index=list('abcde'))
print('Series with NaN:')
print(s_missing)
print()
print('Is null?')
print(s_missing.isnull())
print()
print('After filling NaN with 0:')
print(s_missing.fillna(0))
print()
print('After dropping NaN:')
print(s_missing.dropna())
```

# PART 2: 📋 Pandas DataFrame
A **DataFrame** is a 2D labeled data structure — like a spreadsheet or SQL table.

## 2.1 Creating a DataFrame

```python
# From a dictionary of lists
df = pd.DataFrame({
    'Name':       ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank'],
    'Age':        [25, 30, 35, 28, 22, 40],
    'City':       ['Chennai', 'Mumbai', 'Delhi', 'Bangalore', 'Chennai', 'Pune'],
    'Salary':     [45000, 70000, 85000, 62000, 38000, 95000],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'HR', 'Finance'],
    'Experience': [2, 5, 10, 4, 1, 15]
})
df
```

```python
# From list of dictionaries (each dict = one row)
df2 = pd.DataFrame([
    {'Product': 'Laptop',  'Price': 55000, 'Stock': 10},
    {'Product': 'Phone',   'Price': 20000, 'Stock': 25},
    {'Product': 'Tablet',  'Price': 30000, 'Stock': 15},
])
df2
```

## 2.2 Exploring a DataFrame

```python
print('Shape:', df.shape)          # (rows, columns)
print('Columns:', df.columns.tolist())
print('dtypes:\n', df.dtypes)
print()
print('First 3 rows:')
print(df.head(3))
print()
print('Last 2 rows:')
print(df.tail(2))
```

```python
# Statistical summary
df.describe()
```

```python
# Info about the DataFrame
df.info()
```

## 2.3 Selecting Data

```python
# Select a single column → returns Series
print(df['Name'])
print(type(df['Name']))
```

```python
# Select multiple columns → returns DataFrame
df[['Name', 'Salary', 'Department']]
```

```python
# .loc  — label-based selection
print('Row 0 to 2, specific columns:')
print(df.loc[0:2, ['Name', 'Age', 'Salary']])
```

```python
# .iloc — position-based selection
print('First 3 rows, first 3 columns:')
print(df.iloc[:3, :3])
```

## 2.4 Filtering Rows

```python
# Single condition
print('Employees in IT department:')
print(df[df['Department'] == 'IT'])
```

```python
# Multiple conditions  (& = AND,  | = OR)
print('Salary > 60000 AND Age < 35:')
print(df[(df['Salary'] > 60000) & (df['Age'] < 35)])
```

```python
# .isin() filter
print('Employees in Chennai or Pune:')
print(df[df['City'].isin(['Chennai', 'Pune'])])
```

```python
# String filtering
print('Names starting with A, B, or C:')
print(df[df['Name'].str.startswith(('A', 'B', 'C'))])
```

## 2.5 Adding & Modifying Columns

```python
# Add a new column
df['Annual_Salary'] = df['Salary'] * 12

# Derived column
df['Senior'] = df['Experience'] >= 10

# Apply a function
df['Grade'] = df['Salary'].apply(lambda x: 'High' if x > 70000 else ('Mid' if x > 50000 else 'Low'))

df
```

## 2.6 Sorting

```python
# Sort by single column
print('Sorted by Salary (desc):')
print(df[['Name','Salary']].sort_values('Salary', ascending=False))
```

```python
# Sort by multiple columns
print('Sorted by Department then Salary:')
print(df[['Name','Department','Salary']].sort_values(['Department','Salary'], ascending=[True, False]))
```

## 2.7 Handling Missing Values

```python
# Create a DataFrame with missing values
df_miss = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, np.nan, 5],
    'C': ['x', 'y', 'z', None, 'w']
})
print('With NaN:')
print(df_miss)
print()
print('Missing count per column:')
print(df_miss.isnull().sum())
```

```python
# Fill missing values
print('Fill with mean:')
print(df_miss['A'].fillna(df_miss['A'].mean()))

# Drop rows with any NaN
print('\nDrop rows with NaN:')
print(df_miss.dropna())
```

## 2.8 GroupBy & Aggregation

```python
# Average salary by department
print('Average Salary by Department:')
print(df.groupby('Department')['Salary'].mean())
```

```python
# Multiple aggregations
print('Department summary:')
df.groupby('Department').agg(
    Headcount=('Name', 'count'),
    Avg_Salary=('Salary', 'mean'),
    Max_Salary=('Salary', 'max'),
    Avg_Age=('Age', 'mean')
).round(2)
```

```python
# Count employees per city
print('Employees per City:')
print(df['City'].value_counts())
```

## 2.9 Merging & Concatenating DataFrames

```python
# Two DataFrames to merge
employees = pd.DataFrame({
    'EmpID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'DeptID': [10, 20, 10, 30]
})
departments = pd.DataFrame({
    'DeptID': [10, 20, 30],
    'DeptName': ['HR', 'IT', 'Finance']
})

# Inner merge (like SQL JOIN)
merged = pd.merge(employees, departments, on='DeptID', how='inner')
print('Merged DataFrame:')
print(merged)
```

```python
# Concatenate (stack rows)
df_a = pd.DataFrame({'X': [1,2,3], 'Y': [4,5,6]})
df_b = pd.DataFrame({'X': [7,8,9], 'Y': [10,11,12]})

result = pd.concat([df_a, df_b], ignore_index=True)
print('Concatenated:')
print(result)
```

## 2.10 Pivot Tables

```python
pivot = df.pivot_table(
    values='Salary',
    index='Department',
    columns='Grade',
    aggfunc='mean',
    fill_value=0
).round(0)

print('Pivot — Avg Salary by Dept & Grade:')
pivot
```

```python
# ── DataFrame Tasks ──
s = students.copy()

# Task 1
s['Total']   = s['Maths'] + s['Science'] + s['English']
s['Average'] = (s['Total'] / 3).round(2)

# Task 2
s['Result'] = s.apply(lambda r: 'Pass' if r['Maths']>=70 and r['Science']>=70 and r['English']>=70 else 'Fail', axis=1)

# Task 3
top3 = s[['Name','Class','Total']].sort_values('Total', ascending=False).head(3)
print('Top 3 Students:')
print(top3)

# Task 4
print('\nClass Averages:')
print(s.groupby('Class').agg(Avg_Maths=('Maths','mean'), Avg_Science=('Science','mean'), Count=('Name','count')))

# Task 5
filtered = s[((s['Maths']>85)|(s['Science']>85)) & (s['Absent_Days']<=3)]
print('\nFiltered students:')
print(filtered[['Name','Maths','Science','Absent_Days']])

# Task 6
s['Rank'] = s['Total'].rank(ascending=False, method='min').astype(int)
print('\nLeaderboard:')
print(s[['Rank','Name','Total','Average','Result']].sort_values('Rank'))
```

## 🎯 Quick Reference Cheat Sheet

| Operation | Series | DataFrame |
|---|---|---|
| Create | `pd.Series([...])` | `pd.DataFrame({...})` |
| Select | `s['label']` / `s.iloc[0]` | `df['col']` / `df.loc[]` / `df.iloc[]` |
| Filter | `s[s > 10]` | `df[df['col'] > 10]` |
| Sort | `s.sort_values()` | `df.sort_values('col')` |
| Stats | `s.mean()`, `s.sum()` | `df.describe()` |
| Missing | `s.fillna()`, `s.dropna()` | `df.fillna()`, `df.dropna()` |
| Group | — | `df.groupby('col').agg(...)` |
| Merge | — | `pd.merge(df1, df2, on='key')` |
| Stack | `pd.concat([s1,s2])` | `pd.concat([df1,df2])` |
