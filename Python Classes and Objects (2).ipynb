```python
class Dog:

    def sound(self):

        print("Dog barks")

class Cat:

    def sound(self):

        print("Cat meows")

# Function that works with any object

def make_sound(animal):

    animal.sound()

# Create objects

dog = Dog()

cat = Cat()

# Call function

make_sound(dog)

make_sound(cat)




```

```python
class Bird:

    def fly(self):

        print("Bird can fly")

class Airplane:

    def fly(self):

        print("Airplane can fly")

# Create objects

b = Bird()

a = Airplane()

# Same method name, different behavior

b.fly()

a.fly()




```
