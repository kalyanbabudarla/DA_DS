```python
#oprators

a = np.array ([1,2,3,4])
b = np.array ([4,5,3,4])
print (a+b)
print (a*b)
```

```python
#data types array

a = np.array ([1,2,3,4], dtype=float)
print(a)
```

```python
#math function

a = np.array ([1,2,3,4])
print(np.square(a))
```

```python
#statics function

a = np.array([1,2,3,4,5,4,5])
print(np.mean(a))     # average
print(np.median(a))
print(np.std(a))      # standard deviation
print(np.sum(a))
print(np.min(a))
print(np.max(a))
# print(np.mod(a))
```
