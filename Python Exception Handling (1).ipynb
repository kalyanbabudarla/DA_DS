```python
#Task 1: Square of a Number with Invalid Input Handling

try:
    num = int(input("Enter a number: "))
    print("Square:", num ** 2)
except ValueError:
    print("Invalid input! Please enter a numeric value.")
```

```python
#Task 2: Division with else Block

try:
    a = int(input("Enter numerator: "))
    b = int(input("Enter denominator: "))
    result = a / b
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
except ValueError:
    print("Invalid input! Enter numbers only.")
else:
    print("Division result:", result)
```

```python
#Task 3: Age Validation

age = int(input("Enter your age: "))
if age < 18:
    raise ValueError("Age must be at least 18.")
else:
    print("Access granted.")
```

```python
#Task 4: Custom Bank Exception

class InsufficientFundsError(Exception):
    pass

balance = 5000
withdraw = int(input("Enter withdrawal amount: "))

if withdraw > balance:
    raise InsufficientFundsError("Withdrawal exceeds available balance!")
else:
    balance -= withdraw
    print("Withdrawal successful. Remaining balance:", balance)
```

```python
#Task 5: File Handling with Multiple Exceptions

try:
    with open("sample.txt", "r") as f:
        content = f.read()
        print(content)
except FileNotFoundError:
    print("Error: File not found.")
except PermissionError:
    print("Error: Permission denied.")
except Exception as e:
    print("An unexpected error occurred:", e)
```

```python
#Task 6: Login System with Exception

class LoginError(Exception):
    pass

username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin" and password == "1234":
    print("Login successful!")
else:
    raise LoginError("Invalid credentials. Access denied.")
```

```python
# program to print the reciprocal of even numbers

try:
    num = int(input("Enter a number: "))
    assert num % 2 == 0
except:
    print("Not an even number!")
else:
    reciprocal = 1/num
    print(reciprocal)
```

```python
# Grandparent class

class Animal:

    def eat(self):

        print("Animal eats")

# Parent class

class Dog(Animal):

    def bark(self):

        print("Dog barks")

# Child class

class Puppy(Dog):

    def cry(self):

        print("Puppy cries")

# Using the classes

p = Puppy()

p.eat()   # from Animal

p.bark()  # from Dog

p.cry()   # from Puppy




```
