# conditional statement


```python
# if statement

age = 18
if (age) >= 18:
    print("Eligible to vote")
```

```python
# if else statement

age = 10
if (age) >= 18:
    print("Eligible to vote")
else:
    print("Not for eligible to vote")
```

```python
# Elif statement

age = 20
if age <= 12:
    print("Child")
elif age <= 19:
    print("Teenager")
elif age <= 35:
    print("Young adult")
else:
    print("adult")
```

```python
#  statement

age = 80
is_member = True #False

if age >= 60:
    if is_member:
        print("30% senior discount!")
    else:
        print("20% senior discount.")
else:
    print("Not eligible for a senior discount.")
```
