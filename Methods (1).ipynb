```python
#Non key word argguments
def fun(*a):
    return max(a)
fun(2,4,8)
#print(a)
```

```python
def multiply_all(*numbers):
    total = 1
    for n in numbers:
        print(total)
        total *= n
        print(total)
    return total
multiply_all(2,5,7)
```

```python
#key word argguments
def fun(**kwargs):
    for w, val in kwargs.items():
        print(w, " = ", val)  

fun(s1='Python', s2='and', s3='Jupyther')
```

```python
def kal(**x):
    for y, babu in x.items():
        print(y, " = ", babu)

kal(s1 = "ice", s2 = "ball", s3 = "car")
```

```python
def disp(**kwargs):

    for key , v in kwargs.items():

        print(key,"  : ", v)

disp(name ="abc", age =  30 , city = "chennai")
```
